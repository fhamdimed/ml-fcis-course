# Part IV — Backpropagation

For sigmoid output with binary cross-entropy:

$$
\delta_o
=
\frac{\partial L}{\partial z_o}
=
p-y.
$$

The output-layer gradients are:

$$
\frac{\partial L}{\partial v_j}
=
\delta_o a_j
$$

and:

$$
\frac{\partial L}{\partial b_o}
=
\delta_o.
$$

For hidden unit $j$:

$$
\delta_j
=
\delta_o v_j \sigma'(z_j)
$$

with:

$$
\sigma'(z_j)=a_j(1-a_j).
$$

# Part VI — A Practical Neural-Network Classifier

We now train a small multilayer perceptron on a nonlinear two-class dataset.

We use the **two moons** dataset because its curved class boundary cannot be represented well by a single straight line.

We will compare:

- a small hidden layer;
- a larger hidden layer.

The goal is not to prove that a larger network is always better.

````python
X_all, y_all = make_moons(
    n_samples=700,
    noise=0.22,
    random_state=101
)

df_master = pd.DataFrame(
    X_all,
    columns=["x1", "x2"]
)
df_master["class"] = y_all

display(df_master.head())
````

## Task 6.2 — Predict Model Difficulty

Before training:

1. Would logistic regression with one linear boundary fit this structure perfectly?
2. Why can a hidden layer help?

````python
small_net = Pipeline(steps=[
    ("scale", StandardScaler()),
    ("mlp", MLPClassifier(
        hidden_layer_sizes=(4,),
        activation="relu",
        solver="adam",
        learning_rate_init=0.01,
        max_iter=1200,
        random_state=SEED
    ))
])

large_net = Pipeline(steps=[
    ("scale", StandardScaler()),
    ("mlp", MLPClassifier(
        hidden_layer_sizes=(20, 20),
        activation="relu",
        solver="adam",
        learning_rate_init=0.01,
        max_iter=1200,
        random_state=SEED
    ))
])

small_net.fit(X_train, y_train)
large_net.fit(X_train, y_train)

def classification_summary(model, X_train, y_train, X_valid, y_valid):
    train_pred = model.predict(X_train)
    valid_pred = model.predict(X_valid)
    valid_prob = model.predict_proba(X_valid)[:, 1]

    return {
        "train_accuracy": accuracy_score(y_train, train_pred),
        "valid_accuracy": accuracy_score(y_valid, valid_pred),
        "valid_f1": f1_score(y_valid, valid_pred),
        "valid_log_loss": log_loss(y_valid, valid_prob),
    }

comparison = pd.DataFrame({
    "Small network": classification_summary(
        small_net, X_train, y_train, X_valid, y_valid
    ),
    "Larger network": classification_summary(
        large_net, X_train, y_train, X_valid, y_valid
    ),
}).T

display(comparison.round(4))
````

# Part VII — Decision Boundary Visualization

We will visualize the network's learned class boundary.

````python
def plot_decision_boundary(model, X_data, y_data, title):
    x1_min, x1_max = X_data["x1"].min() - 0.4, X_data["x1"].max() + 0.4
    x2_min, x2_max = X_data["x2"].min() - 0.4, X_data["x2"].max() + 0.4

    xx1, xx2 = np.meshgrid(
        np.linspace(x1_min, x1_max, 260),
        np.linspace(x2_min, x2_max, 260)
    )

    grid = pd.DataFrame({
        "x1": xx1.ravel(),
        "x2": xx2.ravel()
    })

    probs = model.predict_proba(grid)[:, 1].reshape(xx1.shape)

    plt.figure(figsize=(7, 5))
    plt.contour(
        xx1, xx2, probs,
        levels=[0.5],
        linewidths=2
    )
    plt.scatter(
        X_data["x1"],
        X_data["x2"],
        c=y_data,
        alpha=0.7
    )
    plt.xlabel("x1")
    plt.ylabel("x2")
    plt.title(title)
    plt.show()

plot_decision_boundary(
    small_net,
    X_train,
    y_train,
    "Small Neural Network Decision Boundary"
)
````

# Lab 10 Summary

You should now be able to explain and compute one full neural-network training step.

### Forward propagation

$$
z^{(1)}=W^{(1)}x+b^{(1)}
$$

$$
a^{(1)}=\sigma(z^{(1)})
$$

$$
z_o=v^\top a^{(1)}+b_o
$$

$$
p=\sigma(z_o)
$$

### Binary cross-entropy

$$
L(y,p)
=
-\left[
y\log(p)+(1-y)\log(1-p)
\right]
$$

### Backpropagation

$$
\delta_o=p-y
$$

$$
\delta_j
=
\delta_o v_j a_j(1-a_j)
$$

### Parameter update

$$
\theta
\leftarrow
\theta-\alpha\nabla_\theta L.
$$

### Main lessons

- neural networks compose many simple differentiable units;
- nonlinear activations make multilayer models expressive;
- backpropagation applies the chain rule efficiently;
- every weight receives a gradient based on its contribution to the loss;
- larger networks are not automatically better;
- validation evidence remains essential.
